### Get text by tokenks/words

```python
txt = nltk.corpus.gutenberg.words('blake-poems.txt')
len(txt) 
```

Get stopwords set

```python
from nltk.corpus import stopwords
sw = set(stopwords.words('english'))
```

### Remove punctuation tokens from text

```python
import string
txt = [w for w in txt if w not in string.punctuation]
len(txt)
```

But there are still some weird characters and tokens in the text:

```python
'."' in txt
```

It's better to use regular expressions

```python
from nltk.tokenize import RegexpTokenizer

tokenizer = RegexpTokenizer(r'\w+')
txt = tokenizer.tokenize(' '.join(txt))
len(txt)  # removes more words
```

### Filter non-stopwords and also keep them in lowercase

```python
filtered = [w.lower() for w in txt if w.lower() not in sw]
len(filtered)
```

### Count word occurencies and sort by frequency

```python
from collections import Counter

cnt = Counter(filtered).most_common(25)
cnt
```
