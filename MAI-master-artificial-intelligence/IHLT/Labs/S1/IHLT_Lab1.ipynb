Develop a jupyter notebook that show the 25 non-stopwords with more number of occurrences in the file 'blake-poems.txt' of Gutenberg corpus.

```python
import nltk

nltk.download('gutenberg')
nltk.download('stopwords')
```

```python
import re

from collections import Counter

from nltk.corpus import gutenberg, stopwords
from nltk.text import Text
```



1.   Obtain the blake poems corpus.
2.   Filter stopwords.
3.   Filter punctuation symbols
(We don't use *string.punctuation* as there are also strange characters as punctuation symbol combinations *."*).

**Note:** We consider in this problem that uppercase and lowercase words as the same.


```python
corpus = Text(gutenberg.words('blake-poems.txt'))

sw = stopwords.words('english')

filtered_corpus = [w.lower() for w in corpus
                   if w.lower() not in sw                  # Remove stopwords
                   and re.search(r'^\w+$', w) is not None] # Remove punctuation and strange symbols
```

Sort by occurrence using built-in Counter collection and print first 25

```python
counter = Counter(filtered_corpus)

for idx, (w, occ) in enumerate(counter.most_common(25)):
    print(f'#{idx + 1:2}: {w:8} ({occ})')
```
